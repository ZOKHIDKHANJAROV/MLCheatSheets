3. Закодируйте `education` числами в алфавитном порядке: Bachelor = 0, Master = 1, PhD = 2 (как `LabelEncoder`).

## Подсказки

- Сначала посчитайте μ и σ вручную через NumPy, а потом сверьтесь со `StandardScaler` — числа должны совпасть.
- У pandas `df['age'].std()` по умолчанию считает с ddof = 1, а у NumPy и `StandardScaler` — с ddof = 0. Разница видна уже в третьем знаке.
- `LabelEncoder` сортирует категории по алфавиту: `sorted(df['education'].unique())` даст порядок кодов.

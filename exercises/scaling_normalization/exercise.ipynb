## Задание

Дан файл `input.csv` с данными о людях: `height`, `weight`, `age`, `income`. Примените к каждому признаку свой метод масштабирования:

1. **MinMaxScaler** для `height`: x′ = (x − min) / (max − min) — диапазон [0, 1].
2. **StandardScaler** для `weight`: x′ = (x − mean) / std, std с ddof = 0.
3. **RobustScaler** для `age`: x′ = (x − median) / IQR, где IQR = Q3 − Q1 (квантили 25 % и 75 % с линейной интерполяцией, как `np.percentile` по умолчанию).
4. **MaxAbsScaler** для `income`: x′ = x / max(|x|).

## Подсказки

- Стандартное отклонение — с ddof = 0 (у pandas `.std()` по умолчанию ddof = 1).
- Для IQR: `np.percentile(x, 75) - np.percentile(x, 25)`. Подумайте, какой из четырёх методов меньше всего пострадает, если добавить человека с доходом 10 000 000.
